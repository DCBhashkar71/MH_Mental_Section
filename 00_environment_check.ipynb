# 00 — Environment Check

Run every cell below, top to bottom (`Shift+Enter`). This notebook:

1. Confirms all required packages are installed and importable, and prints their versions.
2. Checks whether a GPU is available for the deep-learning notebook later (not required, just informative).

## 1. Package check

```python
import importlib

required_packages = [
    "pandas", "numpy", "scipy", "sklearn", "imblearn",
    "xgboost", "optuna", "shap", "matplotlib", "seaborn",
    "openpyxl", "yaml", "torch",
]

print(f"{'Package':<15}{'Status':<10}{'Version'}")
print("-" * 40)

missing = []
for pkg in required_packages:
    try:
        mod = importlib.import_module(pkg)
        version = getattr(mod, "__version__", "unknown")
        print(f"{pkg:<15}{'OK':<10}{version}")
    except ImportError:
        print(f"{pkg:<15}{'MISSING':<10}-")
        missing.append(pkg)

if missing:
    print(f"\n⚠ Missing packages: {missing}")
    print("Run: pip install -r requirements.txt   (with your 'matmh' environment active)")
else:
    print("\n✅ All required packages are installed.")
```

## 2. GPU check (informative only — CPU is fine for this project)

```python
import torch

if torch.cuda.is_available():
    print(f"✅ GPU available: {torch.cuda.get_device_name(0)}")
elif torch.backends.mps.is_available():
    print("✅ Apple Silicon GPU (MPS) available.")
else:
    print("ℹ No GPU detected — the deep-learning notebook (08) will run on CPU.")
    print("  This is fine: the datasets here are small/medium tabular data, not image/text scale.")
```

## 3. Create required (currently empty) subfolders

```python
import os

# This notebook lives in notebooks/, so the project root is one level up
PROJECT_ROOT = os.path.abspath(os.path.join(os.getcwd(), ".."))
print("Project root detected as:", PROJECT_ROOT)

folders_to_create = [
    "data/interim",
    "data/processed",
    "data/external_validation",
    "models/pakistan",
    "models/arab5",
    "models/uganda",
    "models/ensemble_final",
    "reports/figures",
    "reports/tables",
    "reports/manuscript_drafts",
]

for folder in folders_to_create:
    full_path = os.path.join(PROJECT_ROOT, folder)
    os.makedirs(full_path, exist_ok=True)
    print(f"✅ {folder}")

print("\nAll folders ready.")
```

## 4. Load config and confirm raw files are found

```python
import yaml

config_path = os.path.join(PROJECT_ROOT, "config", "config.yaml")

if not os.path.exists(config_path):
    print(f"⚠ config.yaml not found at {config_path}")
    print("  Save the config.yaml file into your project's config/ folder, then re-run this cell.")
else:
    with open(config_path, "r") as f:
        config = yaml.safe_load(f)
    print("✅ config.yaml loaded.")
    print(f"   Seed: {config['seed']}")

    raw_dir = os.path.join(PROJECT_ROOT, config["paths"]["raw_dir"])
    print(f"\nChecking raw files in {raw_dir}:\n")

    all_found = True
    for label, filename in config["raw_files"].items():
        file_path = os.path.join(raw_dir, filename)
        exists = os.path.exists(file_path)
        status = "✅ found" if exists else "❌ NOT FOUND"
        print(f"  {label:<22} {filename:<35} {status}")
        if not exists:
            all_found = False

    if all_found:
        print("\n✅ All six raw files located. You're ready to move to 01 (EDA notebooks).")
    else:
        print("\n⚠ Some files are missing or named differently than config.yaml expects.")
        print("  Either rename your files in data/raw/ to match the raw_files list in config.yaml,")
        print("  or edit config.yaml's raw_files section to match your actual filenames.")
```
